```python
from elasticsearch import Elasticsearch
from sentence_transformers import SentenceTransformer
```

```python
# 初期化
es = Elasticsearch("http://llm-rag-examples-elasticsearch1:9200")
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
INDEX_NAME = "vector-test"
```

```python
# インデックス作成
if not es.indices.exists(index=INDEX_NAME):
    es.indices.create(
        index=INDEX_NAME,
        body={
            "mappings": {
                "properties": {
                    "text": {"type": "text"},
                    "embedding": {
                        "type": "dense_vector",
                        "dims": 384,  # モデルに合わせる
                        "index": True,
                        "similarity": "cosine"
                    }
                }
            }
        }
    )
```

```python
# 文書登録
docs = [
    {"text": "りんごは赤い果物です"},
    {"text": "富士山は日本一高い山です"},
    {"text": "東京は日本の首都です"},
]
```
